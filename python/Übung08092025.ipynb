## Notenbewertung

```python
punkte = float(input("Geben Sie die punktzahl ein:"))
print (f"Dein punkzahl ist {punkte}")
if 0<punkte<50:
   print("Dein Note ist 6")
elif 50<=punkte<60:
   print("Dein Note ist 5")
elif 60<=punkte<70:
   print("Dein Note ist 4")
elif 70<=punkte<80:
   print("Dein Note ist 3")
elif 80<=punkte<90:
   print("Dein Note ist 2")
elif 90<=punkte<=100:
   print("Dein Note ist 1")
else:
   print("Uengueltige Eingabe")
```

## 8.Zahlensumme berechnen Eingabe:
### Mehrere Zahlen (bis „0“ eingegeben wird) Ausgabe: Summe aller eingegebenen Zahlen

```python
summe = 0
while True:
    zahl =int(input("Bitte geben Sie eine Zahl ein: "))
    print(f"Die Zahl ist {zahl}")
    if zahl == 0:
        break
    summe += zahl
print(f"Die Summe ist {summe}")
```

```python
while True:
    zahl = int(input("Gib eine Zahl ein oder 0 zum Beenden: "))
    if zahl == 0:
        break
    # prüffen ob es eine primzahl ist
    elif zahl > 1:
        for i in range(2, int(zahl**0.5) + 1):
            if zahl % i == 0:
                print(f"{zahl} ist keine Primzahl. Bitte eine andere Zahl eingeben.")
                break
        else:
            print(f"{zahl} ist eine Primzahl.")
    else:
        print("Bitte eine Zahl größer als 1 eingeben.")
```

# Kombination von Kontrollstrukturen

## Primzahlprüfung Eingabe: 
### Eine Zahl Verarbeitung: Schleife zur Prüfung, ob die Zahl nur durch 1 und sich selbst teilbar ist Ausgabe: 
#### „Primzahl“ oder „Keine Primzahl“

```python
# Einfache Primzahlprüfung

zahl = int(input("Bitte gib eine Zahl ein: "))

if zahl < 2:
    print("Keine Primzahl")
else:
    ist_prim = True
    for i in range(2, zahl):
        if zahl % i == 0:
            ist_prim = False
            break
    if ist_prim:
        print("Primzahl")
    else:
        print("Keine Primzahl")
```

### Passwortprüfung mit Wiederholungen Eingabe: 
### Passwort Verarbeitung: Vergleiche mit dem gespeicherten Passwort Schleife:
### Maximal 3 Versuche Ausgabe: Erfolg oder Sperrung

```python
passwort = "weiterbildung "
versuche = 3
for i in range(versuche):
    if input("Bitte geben Sie das Passwort ein:") == passwort:
        print("Das Passwort ist korrekt:Erfolgreich")
        break
    else: 
        print(f"Das Passwort war falsch, noch {versuche -i -1} versuche übrig.")
else:
        
    print("Leider die eingegebene Passwörter sind faslch:Sperrung")
```
