### Aufgabe 1: Größere Zahl finden
Schreibe einen Pseudocode, der zwei Zahlen A und B einliest und ausgibt, welche der beiden Zahlen größer ist – oder ob sie gleich sind.

**Beispiel:**
Eingabe: 5 und 7 → Ausgabe: 7 ist größer als 5.

```python
ALGORITHMUS VergleicheZahlen
    AUSGABE "Gib die erste Zahl ein: "
    EINGABE a
    AUSGABE "Gib die zweite Zahl ein: "
    EINGABE b

    WENN a > b DANN
        AUSGABE a, " ist größer als ", b
    SONST WENN a < b DANN
        AUSGABE b, " ist größer als ", a
    SONST
        AUSGABE "Beide Zahlen sind gleich."
    ENDE WENN
ENDE ALGORITHMUS
```

### Aufgabe 2: Notendurchschnitt berechnen
Schreibe einen Pseudocode, der den Durchschnitt von n eingegebenen Noten berechnet. Der Benutzer soll zuerst angeben, wie viele Noten er eingeben möchte. Danach werden diese Noten eingelesen, summiert und der Durchschnitt ausgegeben.

```python
ALGORITHMUS NotenDurchschnitt
    AUSGABE "Wie viele Noten möchtest du eingeben?"
    EINGABE n

    summe ← 0

    FÜR i VON 1 BIS n WIEDERHOLE
        AUSGABE "Gib Note ", i, " ein:"
        EINGABE note
        summe ← summe + note
    ENDE FÜR

    durchschnitt ← summe / n

    AUSGABE "Der Durchschnitt der ", n, " Noten ist: ", durchschnitt
ENDE ALGORITHMUS
```

### Aufgabe 3: Gerade oder ungerade?
Schreibe einen Pseudocode, der eine Zahl einliest und prüft, ob sie gerade oder ungerade ist.

**Beispiel:**
Eingabe: 4 → Ausgabe: Die Zahl ist gerade.

```python
ALGORITHMUS GeradeOderUngerade
    AUSGABE "Gib eine Zahl ein:"
    EINGABE zahl

    WENN zahl MODULO 2 = 0 DANN
        AUSGABE "Die Zahl ist gerade."
    SONST
        AUSGABE "Die Zahl ist ungerade."
    ENDE WENN
ENDE ALGORITHMUS
```

### Aufgabe 4: Kleinste Zahl finden
Schreibe einen Pseudocode, der drei Zahlen einliest und die kleinste davon ausgibt.

**Beispiel:**
Eingabe: 5, 2, 8 → Ausgabe: Die kleinste Zahl ist 2.

```python
ALGORITHMUS KleinsteVonDrei
    AUSGABE "Gib die erste Zahl ein:"
    EINGABE a
    AUSGABE "Gib die zweite Zahl ein:"
    EINGABE b
    AUSGABE "Gib die dritte Zahl ein:"
    EINGABE c

    kleinste ← a

    WENN b < kleinste DANN
        kleinste ← b
    ENDE WENN

    WENN c < kleinste DANN
        kleinste ← c
    ENDE WENN

    AUSGABE "Die kleinste Zahl ist ", kleinste
ENDE ALGORITHMUS
```

### Aufgabe 5: Wechselgeld berechnen
Ein Kunde bezahlt mit einem bestimmten Geldbetrag. Schreibe einen Pseudocode, der berechnet, wie viele 2-Euro-, 1-Euro- und 50-Cent-Münzen als Wechselgeld ausgegeben werden sollen. Hinweis Modulo.

**Beispiel:**
Eingabe: Preis = 7.50 €, Bezahlt = 10.00 € → Ausgabe: 1 × 2-Euro-Münze, 1 × 50-Cent-Münze.

```python
ALGORITHMUS WechselgeldBerechnung
    AUSGABE "Gib den Preis ein (in Euro):"
    EINGABE preis
    AUSGABE "Gib den bezahlten Betrag ein (in Euro):"
    EINGABE bezahlt

    wechselgeld ← bezahlt - preis
    wechselgeld_cent ← wechselgeld * 100    // Umrechnung in Cent

    anzahl2Euro ← wechselgeld_cent DIV 200
    rest ← wechselgeld_cent MOD 200

    anzahl1Euro ← rest DIV 100
    rest ← rest MOD 100

    anzahl50Cent ← rest DIV 50
    rest ← rest MOD 50

    AUSGABE "Wechselgeld: " AUS
    WENN anzahl2Euro > 0 DANN
        AUSGABE anzahl2Euro, " x 2-Euro-Münze(n)"
    ENDE WENN

    WENN anzahl1Euro > 0 DANN
        AUSGABE anzahl1Euro, " x 1-Euro-Münze(n)"
    ENDE WENN

    WENN anzahl50Cent > 0 DANN
        AUSGABE anzahl50Cent, " x 50-Cent-Münze(n)"
    ENDE WENN
ENDE ALGORITHMUS
```
