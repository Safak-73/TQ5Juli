```python
class SkyportBaggageSystem:
    def assignBaggageToFlights(self, baggage_list, flight_list):
        # Erstelle ein Dictionary zur schnellen Flugnummer-Suche
        flight_map = {flight.flightNumber: flight for flight in flight_list}

        # Liste für nicht zugewiesene Gepäckstücke
        unassigned_baggage = []

        # Durchlaufe alle Gepäckstücke
        for baggage in baggage_list:
            flight = flight_map.get(baggage.destinationFlight)
            if flight:
                flight.addBaggage(baggage)
            else:
                unassigned_baggage.append(baggage)

        # Ausgabe
        for flight in flight_list:
            print(f"Flug {flight.flightNumber} hat {len(flight.assignedBaggage)} Gepäckstücke:")
            for b in flight.assignedBaggage:
                print(f"  {b.baggageId} ({b.passengerName})")

        for b in unassigned_baggage:
            print(f"Gepäck {b.baggageId} konnte nicht zugewiesen werden.")
```
