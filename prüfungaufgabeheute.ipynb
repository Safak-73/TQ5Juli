```python
def berechne_rabatt(betrag):
    if betrag < 100:
        rabatt_prozent = 0
    elif betrag < 500:
        rabatt_prozent = 5
    elif betrag < 1000:
        rabatt_prozent = 10
    else:
        rabatt_prozent = 15
    rabatt_betrag = betrag * rabatt_prozent / 100
    endbetrag = betrag - rabatt_betrag


    print(f"Einkaufswert: {betrag:.2f} €")
    print(f"Rabatt: {rabatt_prozent} %")
    print(f"Endbetrag: {endbetrag:.2f} €")

    return endbetrag
```

```python
berechne_rabatt(80)     # 0 % Rabatt
berechne_rabatt(300)    # 5 % Rabatt
berechne_rabatt(750)    # 10 % Rabatt
berechne_rabatt(1200)
```
