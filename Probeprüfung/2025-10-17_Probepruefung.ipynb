## Aufgabe 3:
### Ausgangssituation
Ein Programm soll ein eingegebenes Wort zeichenweise rückwärts ausgeben. Es dürfen keine sprachspezifischen Komfortfunktionen zur Umkehr von Strings oder Arrays verwendet werden.

### Ihre Aufgabe Schreiben Sie ein Programm, das: 
1. Ein Wort (ohne Leerzeichen) von der Konsole einliest. 
2. Das Wort ohne Hilfsfunktionen rückwärts ausgibt. 
3. Nur Schleifen und Indexzugriffe nutzt (z. B. for, while), sowie elementare Operationen auf Zeichen/Arrays. 

### Rahmenbedingungen / Verbote 
- Verboten: eingebaute Reverse-Funktionen, String-Slicing/Range-Operatoren wie [::-1], Reverse(), Collections.reverse, StringBuilder.reverse(), LINQ, Streams o. Ä. 
- Erlaubt: Schleifen, Zuweisungen, einfache Variablen, Zeichen-/Arrayzugriffe. 

### Ein-/Ausgabeformat 
- Eingabe: Ein Wort, z. B. Programmieren 
- Ausgabe: Rückwärts, z. B. neriemmargorP

(15 Punkte)

```python
s = input().strip()          # z.B. "Programmieren"
for i in range(len(s) - 1, -1, -1):
    print(s[i], end='')
print()
```
