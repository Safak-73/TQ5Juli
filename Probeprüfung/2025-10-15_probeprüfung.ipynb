3. Die anzufliegenden Positionen liegen in eine Liste geoPositions als Objekte der Klasse GeoPos vor. GeoPos hat die folgende attribute: latitude: float, longitude: float, altitude: float.
4. Für Entfernungsberechnungen zwischen Geopositionen kann die Methode getDistance der Klasse GeoCalculator verwendet werden: def getDistance(pos1: GeoPos, pos2: GeoPos)
5. Der Algorithmus für den Drohnenflug soll sein Ergebnis in eine Liste flightPositions speichern und zurückgeben. Die Arbeitsweise des Algorithmus calculateFlight() wird wie folgt beschrieben:

- Die Liste geoPositions kann durch den Algorithmus verändert werden.
- Die Geoposition mit dem Index 0 der Liste geoPositions wird zur ersten aktuellen Position. Diese Position wird in die Liste flightPositions gespeichert (Startposition!) und kann dann aus der Liste der noch anzufliegenden Positionen (geoPositions) gelöscht werden. Hinweis: bereits implementiert, siehe Vorgabe.
- Solange noch Geopositionen im Array geoPositions vorhanden sind: Ermitteln der Position im Array geoPositions mit der kürzesten Entfernung zur aktuellen Position. Diese Position wird im Array flightPositions gespeichert und zur neuen aktuellen Position.  Diese Position kann dann vom Array geoPositions gelöscht werden. 

Hinweis: Der größte float wert kann mit float.MAX_VALUE abgerufen werden.
Vervollständigen Sie dahingehend den Pseudocode:

(Punkte 20)

````python
calculateFlight(geoPositions: GeoPos[] ) : GeoPos[]
    flight_positions = new GeoPos[geoPositions.length]          # Array der angeflogenen Positionen
    flightPositions[0] = geoPositions[0]                        # Position 0 als Startposition
    currentPosition = geoPositions[0]
    geoPositions.remove(0)                                      # aus dem Array der noch anzufliegenden Positionen entfernen

    WHILE geoPositions.length > 0 DO
        shortestDistance = float.MAX_VALUE
        nearestPosition = null
        FOR position IN geoPositions DO
            distance = currentPosition.distanceTo(position)     # Distanz zur aktuellen Position berechnen
            IF distance < shortestDistance THEN
                shortestDistance = distance
                nearestPosition = position
            ENDIF
        ENDFOR

        flightPositions.add(nearestPosition)                    # Nächste Position zur Flugroute hinzufügen
        currentPosition = nearestPosition
        geoPositions.remove(nearestPosition)                    # Aus dem Array der noch anzufliegenden Positionen entfernen
    ENDWHILE

    RETURN flightPositions
END
````

````python
def berechne_rabatt(betrag):
    rabatt_prozent = 0

    # Rabattbedingungen ergänzen
    if betrag < 100:
        rabatt_prozent = 0
    elif betrag < 500:
        rabatt_prozent = 5
    elif betrag < 1000:
        rabatt_prozent = 10
    else:
        rabatt_prozent = 15
    
    # Endbetrag nach Rabatt berechnen
    endbetrag = betrag * (1 - rabatt_prozent / 100)

    # Ausgabe der Ergebnisse
    print(f"Einkaufsbetrag: {betrag} Euro")
    print(f"Rabatt: {rabatt_prozent}%")
    print(f"Endbetrag nach Rabatt: {endbetrag:.2f} Euro")

    return endbetrag

# Test
# berechne_rabatt(80)
# berechne_rabatt(300)
# berechne_rabatt(750)
# berechne_rabatt(1200)
````
