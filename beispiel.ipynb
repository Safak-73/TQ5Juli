```python
def erstelle_brett():
    brett = []
    
    for i in range(3):
     zeile = [" "," "," "]
     brett.append(zeile)
    return brett   

def drucke_brett(brett):
   for zeile in brett:
      print("|". join(zeile))
      print("------")
drucke_brett(erstelle_brett())

def zug_machen(brett,spieler,zeile,spalte):
   if brett[zeile][spalte]== " ":
      brett[zeile][spalte]=spieler
      return True
   else:
      return False   

def spiele_tic_tac_toe():
   brett =erstelle_brett()
   akteul_spiele = "x"
   while True:
      drucke_brett(brett)
      try:
        zeile =int(input(f"spieler {akteul_spiele}, wäle deine zeile(0-2)"))
        spalte =int(input(f"spieler {akteul_spiele}, wäle deine zeile(0-2)"))
      except:
           print('ungultig ,versuch erneut')
           
           continue
      if zeile not in [0,1,2] or spalte not in [0,1,2]:
            print("Ungültige Eingabe! Nur 0, 1 oder 2 erlaubt.")
            continue  
      if not zug_machen(brett, akteul_spiele,zeile, spalte):
            print("Feld schon belegt, versuch nochmal!")
            continue
   
      aktueller_spieler = "O" if aktueller_spieler == "X" else "X"
       

spiele_tic_tac_toe()
      

```
