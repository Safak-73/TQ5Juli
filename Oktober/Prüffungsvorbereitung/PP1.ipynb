```python
def berechne_rabatt(betrag, rabatt_prozent):

        rabatt_prozent = 0

        if betrag < 100:
            rabatt_prozent = 0
        elif betrag > 100 and betrag < 499:
            rabatt_prozent = 5
        elif betrag > 500 and betrag < 999:
            rabatt_prozent = 10
        elif betrag >= 1000:
            rabatt_prozent = 15
        else:
            print("Ungültiger Betrag")
    
        rabatt = betrag * rabatt_prozent / 100
        endbetrag = betrag - rabatt
    
        print(f"Einkaufswert: {betrag} €")
        print(f"Rabatt: {rabatt_prozent} %")
        print(f"Endbetrag: {endbetrag} €")

berechne_rabatt(80, 0)
berechne_rabatt(300, 0)
berechne_rabatt(750, 0)
berechne_rabatt(1200, 0)
```
