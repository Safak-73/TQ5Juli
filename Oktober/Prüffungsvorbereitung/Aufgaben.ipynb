Aufgabe 1 – Kinokartenverwaltung
Szenario:
Ein kleines Kino möchte sein Kassensystem digitalisieren.
Es soll ein Python-Programm entwickelt werden, das Kinotickets verwaltet und den 
Gesamtumsatz berechnet.
Anforderungen:
1. Erstelle eine Klasse Kinoticket mit folgenden Attributen:
o filmname (String)
o preis (float)
o reihe (int)
o platz (int)
2. Implementiere in der Klasse eine Methode:
def anzeigen(self):
# gibt Filminfo, Reihe und Platz formatiert aus
3. Schreibe eine weitere Klasse KinoVerwaltung, die:
o eine Liste von verkauften Tickets speichert,
o eine Methode ticket_hinzufügen(ticket) enthält,
o eine Methode gesamtumsatz() besitzt, die alle Ticketpreise summiert.
4. Erstelle im Hauptprogramm (main) mehrere Tickets, füge sie der Verwaltung hinzu 
und gib am Ende den Gesamtumsatz aus.
Beispielhafte Konsolenausgabe:
Ticketverkauf:
Film: Dune 2, Reihe: 3, Platz: 5, Preis: 9.5 €
Film: Avatar, Reihe: 4, Platz: 7, Preis: 10.0 €
Gesamtumsatz: 19.5 €

```python
class Kiniticket:
# diese klasse enthält die Attribute eines Kinotickets filmname (String) preis (float) reihe (int) platz (int)

    def __init__(self, filmname: str, preis: float, reihe: int, platz: int):
        self.filmname = filmname
        self.preis = preis
        self.reihe = reihe
        self.platz = platz

# Methoden gibt Filminfo, Reihe und Platz formatiert aus

    def anzeigen(self):
        print(f"Film: {self.filmname}, Reihe: {self.reihe}, Platz: {self.platz}, Preis: {self.preis:.2f} €")

class KinoVerwaltung:

    # eine Liste von verkauften Tickets speichert
    def __init__(self):
        self.verkaufte_tickets = []

# eine Methode ticket_hinzufügen(ticket) enthält
    def ticket_hinzufügen(self, ticket, Kinoticket):
        if isinstance(ticket, Kinoticket):
            self.verkaufte_tickets.append(ticket)
        else:
            print("Fehler: Das übergebene Objekt ist kein Kinoticket.")

# eine Methode gesamtumsatz() besitzt, die alle Ticketpreise summiert.
    def gesamtumsatz(self) -> float:
        
        umsatz = sum(ticket.preis for ticket in self.verkaufte_tickets)
        return umsatz


```

```python
def main():

    # Initialisiere die Verwaltung
    verwaltung = KinoVerwaltung()

    # Erstelle mehrere Tickets
    ticket1 = Kinoticket("Dune 2", 9.50, 3, 5)
    ticket2 = Kinoticket("Avatar", 10.00, 4, 7)
    ticket3 = Kinoticket("Oppenheimer", 12.50, 1, 12)

    # Füge die Tickets der Verwaltung hinzu
    verwaltung.ticket_hinzufügen(ticket1)
    verwaltung.ticket_hinzufügen(ticket2)
    verwaltung.ticket_hinzufügen(ticket3)

    print("--- Ticketverkauf ---")

    # Zeige die Details jedes Tickets an
    for ticket in verwaltung.verkaufte_tickets:
        ticket.anzeigen()

    print("\n--- Gesamtbilanz ---")
    # Gesamtumsatz berechnen und ausgeben
    umsatz = verwaltung.gesamtumsatz()
    print(f"Gesamtumsatz: {umsatz:.2f} €")

if __name__ == "__main__":
    main()
```

Aufgabe 2 – Mitarbeiterverwaltung mit Bonusberechnung
Szenario:
Eine Firma möchte die Jahresboni ihrer Mitarbeiter berechnen.
Schreibe ein Python-Programm, das Mitarbeiterdaten verwaltet und den Bonus abhängig 
vom Gehalt und der Dauer der Betriebszugehörigkeit berechnet.
Anforderungen:

1. Erstelle eine Klasse Mitarbeiter mit den Attributen:
o name (String)
o gehalt (float)
o jahre_im_betrieb (int)

2. Implementiere eine Methode:
def berechne_bonus(self):
# 5 % Bonus pro Jahr, maximal 25 % des Gehalts

3. Erstelle eine Klasse Firma, die mehrere Mitarbeiter verwaltet.
o Methode mitarbeiter_hinzufügen(m): fügt Mitarbeiter hinzu
o Methode gesamt_bonus(): summiert alle berechneten Boni

4. Im Hauptprogramm sollen mehrere Mitarbeiter erzeugt und deren Boni sowie der 
Gesamtbonus ausgegeben werden.
Beispielhafte Konsolenausgabe:
Mitarbeiter: Lisa | Gehalt: 4000 € | Jahre: 3 | Bonus: 600 €
Mitarbeiter: Tom | Gehalt: 3500 € | Jahre: 6 | Bonus: 875 €
Gesamtbonus aller Mitarbeiter: 1475 €

```python
class Mitarbeiter:

    def __init__(self, name: str, gehalt: float, jahre_im_betrieb: int):
        self.name = name
        self.gehalt = gehalt
        self.jahre_im_betrieb = jahre_im_betrieb
    
    def berechne_bonus(self) -> float:
        
        jahre = self.jahre_im_betrieb
        
        if jahre >= 5: # 5 Jahre oder mehr 
            return self.gehalt * 0.25 
        elif jahre >= 4: # 4 Jahre -> 20%
            return self.gehalt * 0.20
        elif jahre >= 3: # 3 Jahre -> 15%
            return self.gehalt * 0.15
        elif jahre >= 2: # 2 Jahre -> 10%
            return self.gehalt * 0.10
        elif jahre >= 1: # 1 Jahr -> 5%
            return self.gehalt * 0.05
        else: # Weniger als 1 Jahr
            return 0.0
    
class Firma:

    def __init__(self):
        self.mitarbeiter_liste = []

    def mitarbeiter_hinzufügen(self, mitarbeiter: Mitarbeiter):
        """Fügt einen Mitarbeiter hinzu."""
        self.mitarbeiter_liste.append(mitarbeiter)

    def gesamt_bonus(self) -> float:
        """Summiert alle berechneten Boni."""
        # Ruft berechne_bonus() für jeden Mitarbeiter auf und summiert die Ergebnisse
        gesamtbonus = sum(m.berechne_bonus() for m in self.mitarbeiter_liste)
        return gesamtbonus

def main():
    """
    Hauptprogramm zur Demonstration.
    """
    firma = Firma()

    # Erstelle Mitarbeiter
    # Mitarbeiter 1: 3 Jahre -> 15% von 4000 = 600 €
    mitarbeiter1 = Mitarbeiter("Lisa", 4000.0, 3) 
    # Mitarbeiter 2: 6 Jahre -> max. 25% von 3500 = 875 €
    mitarbeiter2 = Mitarbeiter("Tom", 3500.0, 6) 

    # Füge Mitarbeiter zur Firma hinzu
    firma.mitarbeiter_hinzufügen(mitarbeiter1)
    firma.mitarbeiter_hinzufügen(mitarbeiter2)

    print("--- Jahresboni Berechnung ---")

    # Ausgabe der Mitarbeiterdaten und des berechneten Bonus.
    
    # Mitarbeiter 1 Ausgabe
    bonus1 = mitarbeiter1.berechne_bonus()
    print(f"Mitarbeiter: {mitarbeiter1.name} | Gehalt: {mitarbeiter1.gehalt:.0f} € | Jahre: {mitarbeiter1.jahre_im_betrieb} | Bonus: {bonus1:.0f} €")
    
    # Mitarbeiter 2 Ausgabe
    bonus2 = mitarbeiter2.berechne_bonus()
    print(f"Mitarbeiter: {mitarbeiter2.name} | Gehalt: {mitarbeiter2.gehalt:.0f} € | Jahre: {mitarbeiter2.jahre_im_betrieb} | Bonus: {bonus2:.0f} €")

    # Gesamtbonus berechnen und ausgeben
    gesamtbonus_summe = firma.gesamt_bonus()
    print(f"\nGesamtbonus aller Mitarbeiter: {gesamtbonus_summe:.0f} €")

if __name__ == "__main__":
    main()
```
