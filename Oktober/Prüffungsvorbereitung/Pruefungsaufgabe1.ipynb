```python
class Kinoticket:
    def __init__(self, filmname, preis, reihe, platz):
        self.filmname = filmname
        self.preis = preis
        self.reihe = reihe
        self.platz = platz

#Film: Dune 2, Reihe: 3, Platz: 5, Preis: 9.50 €
    def anzeigen(self):
        print(f'Film: {self.filmname}, Reihe: {self.reihe}, Platz: {self.platz}, Preis: {self.preis:.2f} €')
```

```python
class Kinoverwaltung:
    def __init__(self):
        self.verkaufte_tickets = []

    def ticket_hinzufuegen(self, kinoticket):
        self.verkaufte_tickets.append(kinoticket)

    def gesamtumsatz(self):
        summe = 0
        for ticket in self.verkaufte_tickets:
            summe += ticket.preis
        return summe
```

```python
kino = Kinoverwaltung()

print('Ticketverkauf: ')
kino.ticket_hinzufuegen(Kinoticket('Dune',9.5,3,5))
kino.ticket_hinzufuegen(Kinoticket('Avatar',10,7,4))

for tickets in kino.verkaufte_tickets:
    tickets.anzeigen()

print(f'Gesamtumsatz: {kino.gesamtumsatz():.2f} €')
```

```python
class Firma:
    def __init__(self):
        self.mitarbeiter_liste = []

    def mitarbeiter_hinzufuegen(self, mitarbeiter):
        self.mitarbeiter_liste.append(mitarbeiter)

    def gesamt_bonus(self):
        summe = 0
        for mitarbeiter in self.mitarbeiter_liste:
            summe += mitarbeiter.berechne_bonus()
        return summe
```

```python
class Mitarbeiter:
    def __init__(self, name, gehalt, jahre_im_betrieb):
        self.name = name
        self.gehalt = gehalt
        self.jahre_im_betrieb = jahre_im_betrieb

    def berechne_bonus(self):
        # 5 % Bonus pro Jahr, maximal 25 % des Gehalts
        #Wenn der Gehalt Bonus groesser als 25 ist, dann haben wir ein problem
        #Der Maximum ist 25!
        #Bspw: Mitarbeiter ist 10 Jahre im betrieb:
        #Dann ist der Bonus 10*5 = 50
        #Wir duerfen 50% Bonus aber nicht bekommen! Maximum 25.
        bonus_prozent = self.jahre_im_betrieb * 5
        if bonus_prozent > 25:
            bonus_prozent = 25

        bonus_euro = self.gehalt * (bonus_prozent/100)

        return bonus_euro
```
