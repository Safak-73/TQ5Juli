# Aufgabe 3 – Paketversand-System


Szenario:

Ein Logistikunternehmen möchte ein Programm entwickeln, das Pakete verwaltet, die an 
verschiedene Empfänger verschickt werden.
Das System soll Pakete anlegen, berechnen, welche Kosten entstehen, und alle Daten in einer 
Datei speichern können.

```python
class Paket:
    
    def __init__(self, paketID, gewicht, zielort, versicherung: bool):
        self.paketID = paketID
        self.gewicht = gewicht
        self.zielort = zielort
        self.versicherung = versicherung
    
    def berechne_versandkosten(self):
        basispreis = 5.0
        # + 1 € pro kg über 5 kg
        gewichtszuschlag = max(0, self.gewicht - 5) * 1.0
        # + 2 € wenn versichert = True
        versicherungszuschlag = 2.0 if self.versicherung else 0.0

class Paketverwaltung:

    def __init__(self):
        self.pakete = []
    
    def paket_hinzufuegen(self, paket: Paket):
        self.pakete.append(paket)
    
    def alle_anzeigen(self):
        for paket in self.pakete:
            print(f"PaketID: {paket.paketID}, Gewicht: {paket.gewicht}kg, Zielort: {paket.zielort}, Versicherung: {paket.versicherung}, Versandkosten: {paket.berechne_versandkosten()}€")
            if paket.gewicht < 0:
                print("Warnung: Paketgewicht ist negativ!")
    
    def exportiere_datei(self, dateiname):
        with open(dateiname, 'w') as datei:
            for paket in self.pakete:
                datei.write(f"{paket.paketID},{paket.gewicht},{paket.zielort},{paket.versicherung}\n")

    def gesamt_kosten(self):
        return sum(paket.berechne_versandkosten() for paket in self.pakete) 
    

main = Paketverwaltung()

p1 = Paket("P001", 4.5, "Berlin", True)
p2 = Paket("P002", 6.0, "München", False)
p3 = Paket("P003", -2.0, "Hamburg", True)   
p4 = Paket("P004", 10.0, "Köln", True)

main.paket_hinzufuegen(p1)
main.paket_hinzufuegen(p2)  
main.paket_hinzufuegen(p3)
main.paket_hinzufuegen(p4)

main.alle_anzeigen()

main.gesamt_kosten()

main.exportiere_datei("pakete.txt")
```
