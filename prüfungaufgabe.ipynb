```python
class Kinotickt:
    def __init__(self, filmname: str, preis: float, reihe: int, platz: int):
        self.filmname=filmname
        self.preis=preis
        self.reihe=reihe
        self.platz=platz

    def anzeige(self):
        print ('filminfo')
        print(f"Reihe: {self.reihe}, Platz: {self.platz}, Filmname: {self.filmname}, Preis: {self.preis}")
        
         
```

```python
class KinoVerwaltung:
    def __init__(self):
        self.verkaufte_tickets=[]

    def ticket_hinzufügen(self,ticket:Kinotickt):
           self.verkaufte_tickets.append(ticket)

    def gesammteumsatz(self):
         summe=0
         for ticket in self.verkaufte_tickets:
              summe+=ticket.preis

         
         return summe
    
```

```python
 
def main():
    verwaltung = KinoVerwaltung()
    

    # ticket erstellen
    ticket1= Kinotickt("Inception", 9.50, 5, 12)
    ticket2= Kinotickt("Interstellar", 10.00, 3, 7)
    ticket3= Kinotickt("Tenet", 8.75, 2, 4)

     # ticket hinzufügen

    verwaltung.ticket_hinzufügen(ticket1)
    verwaltung.ticket_hinzufügen(ticket2)
    verwaltung.ticket_hinzufügen(ticket3)

    #ticket anzeigen
    print("verkaufte ticket:")
    for ticket in verwaltung.verkaufte_tickets:
        ticket.anzeige()

    print(f" gesammtumsatz: {verwaltung.gesammteumsatz(): .2f} euro") 

if __name__ == "__main__":
    main()        
```
