Beispielhafte Konsolenausgabe:

Paket-ID: 102 | Gewicht: 2.0 kg | Ziel: Hamburg | Versichert: Nein | Kosten: 7.0 €

Daten erfolgreich in 'pakete.txt' gespeichert.

```python
class Paket:
    def __init__(self, paketid:int, gewicht:float, zielort:str, versichert:bool ):
        self.paketid = paketid
        self.gewicht=gewicht
        self.zielort=zielort
        self.versichert=versichert

    def anzeige(self):
        print('paketinfo')
        print(f"paketid: {self.paketid}, gewicht: {self.gewicht}, zielort: {self.zielort}, versichert: {self.versichert}, kosten: {self.berechne_versandkosten()}")


    def berechne_versandkosten(self):
        # Basispreis 5 €
        # + 1 € pro kg
        # + 2 € wenn versichert = True
        basispreis=5
        basispreis+=self.gewicht*1 
        
        if self.versichert == True:
                    basispreis+=2

        return basispreis






    
```

```python
paket1=Paket(101, 3.5, 'Berlin', 'ja')
paket1.anzeige()
paket1.berechne_versandkosten()
```

```python
class Paketverwaltung:
    def __init__(self):
       self.erstellte_pakete=[]

    def paket_hinzufügen(self, p:Paket):
        if p.gewicht <0:
            print(f"Fehler: Paket mit ID {p.paketid} hat ein ungültiges Gewicht ({p.gewicht} kg).")
            return False
        self.erstellte_pakete.append(p)
        return True

    def alle_anzeigen(self):
        if not self.erstellte_pakete:
            print('keine pakete vorhanden')
        else:
          print('alle gespeicherte pakete')  
          for paket in self.erstellte_pakete:  
              paket.anzeige()
          print(f"Gesamtkosten: {self.gesamt_kosten()} €") 

    
    def  gesamt_kosten(self):
        summe=0
        for paket in self.erstellte_pakete:
            summe+=paket.berechne_versandkosten()
            return summe

   
    def  exportiere_datei(self, dateiname):
        try:
            with open(dateiname,'w') as datei:
                for paket in self.erstellte_pakete:
                     zeile=f"{paket.paketid}, {paket.gewicht}, {paket.zielort}, {paket.versichert}\n"

                     datei.write(zeile)
                print(f"Datei: '{dateiname}'  erfolgreich exportiert.")



        except Exception as e:
            print(f"fehler bei schreiben der datei: {e}") 


    
    def importiere_datei(self, dateiname):
        try:
            with open(dateiname, 'r') as datei:
                for zeile in datei:
                    teile = zeile.strip().split(',')
                    if len(teile) == 4:
                        paketid = int(teile[0])
                        gewicht = float(teile[1])
                        zielort = teile[2]
                        versichert = teile[3].lower() == 'true'
                        paket = Paket(paketid, gewicht, zielort, versichert)
                        self.paket_hinzufügen(paket)
            print(f"Daten erfolgreich aus '{dateiname}' importiert.")
        except FileNotFoundError:
            print(f"Datei '{dateiname}' nicht gefunden.")
        except Exception as e:
            print(f"Fehler beim Einlesen der Datei: {e}")        


```

```python
# Hauptprogramm – NICHT einrücken!
verwaltung = Paketverwaltung()

paket1 = Paket(101, 3.5, "Berlin", True)
paket2 = Paket(102, 2.0, "Hamburg", False)
paket3 = Paket(103, 4.0, "München", True)
paket4 = Paket(104, 1.5, "Köln", False)

verwaltung.paket_hinzufügen(paket1)
verwaltung.paket_hinzufügen(paket2)
verwaltung.paket_hinzufügen(paket3)
verwaltung.paket_hinzufügen(paket4)

verwaltung.alle_anzeigen()
```
