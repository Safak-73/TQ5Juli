Am Flughafen Skyport werden täglich zahlreiche Flüge abgefertigt. Jeder Passagier gibt sein Gepäck am Check-in auf, und das System muss sicherstellen, dass jedes Gepäckstück dem richtigen Flug zugewiesen wird. Das Flughafenmanagement möchte ein automatisches System zur Gepäckzuweisung entwickeln.
Gegebene Klassen: Baggage (Gepäck) und Flight (Flug):

 1. Baggage: 

   * string baggageId - Eindeutige ID des Gepäckstücks
   * string passengerName
   * string destinationFlight - Flugnummer des Ziel-Fluges

```python
class Baggage:
    def __init__(self, baggageId, passengerName, destinationFlight):
        self.baggageId = baggageId
        self.passengerName = passengerName
        self.destinationFlight = destinationFlight
```

2. Flight: 

    * string flightNumber - Eindeutige Flugnummer
    * string destination
    * List assignedBaggage - Liste mit Gepäckstücken
    * Methode: def addBaggage(baggage: Baggage) - Fügt gepäck zu assignedBaggage hinzu.

```python
class Flight:
    def __init__(self,flightNumber,destination):
        self.flightNumber = flightNumber
        self.destination = destination
        self.assignedBaggage = []

    def addBaggage(self,baggage):
        self.assignedBaggage.append(baggage)
```

Aufgabe:


Erstellen Sie eine Methode in der Klasse SkyportBaggageSystem, die die Gepäckstücke den richtigen Flügen zuordnet.

def assignBaggageToFlights(baggage: List, flights: Lists)

 Anforderungen: 

    Jedes Gepäckstück hat eine zugehörige Flugnummer.
    Das System soll das Gepäck dem passenden Flug in der Liste der Flüge zuweisen.
    Falls ein Gepäckstück eine ungültige oder nicht existierende Flugnummer hat, soll es ignoriert werden.
    Ausgabe soll wie im Beispiel unten aussehen.

```python
gepaeck = Baggage('G001','Nico','ABC123')
flug = Flight('ABC123','Karribik')
flug.addBaggage(gepaeck)
flug.assignedBaggage[0].passengerName
```

```python
class SkyportBaggageSystem:
    def assignBaggageToFlights(self,baggage, flights):
        print('Eingabe - Liste der Fluege:\n')
        for flug in flights:
            print(f'Flugnummer: {flug.flightNumber}, Ziel: {flug.destination}')
        print()

        print('Eingabe - Liste der Gepaeckstuecke:\n')
        for gepaeck in baggage:
            print(f'Gepaeck-ID: {gepaeck.baggageId}, Passagier: {gepaeck.passengerName}, Flug: {gepaeck.destinationFlight}')
        print()

        print('Erwartete Ausgabe:')

        for gepaeck in baggage:
            for flug in flights:
                if gepaeck.destinationFlight == flug.flightNumber:
                    flug.addBaggage(gepaeck)
                    break

        for flug in flights:
            print(f'Flug: {flug.flightNumber} hat {len(flug.assignedBaggage)} Gepaeckstuecke: {flug.assignedBaggage}')

        
```

```python
skyportsystem = SkyportBaggageSystem()

fluege = [Flight('AB123','Muenchen'),Flight('XY789','Madrid')]
gepaeckListe = [Baggage('G001','Nicolas','XY789'),Baggage('G002','Bob','AB123'),Baggage('G003','Katherina','BT5T34')]

skyportsystem.assignBaggageToFlights(gepaeckListe,fluege)
```
